## Signal path

The RF-ADCs connect directly to programmable logic. RFSoC-SAM performs
spectral preprocessing, decimation, and FFT processing in programmable logic
before transferring display data to the processing system. Each enabled ADC
channel has an associated spectrum-analyser and flexible-decimator path.


## Hardware configuration

Recorded XM500 loopback mapping:

| Channel | DAC source | ADC destination |
| ---: | --- | --- |
| 0 | DAC4, tile 229 block 0 | ADC0, tile 224 block 0 |
| 1 | DAC5, tile 229 block 1 | ADC1, tile 224 block 1 |
| 2 | DAC6, tile 229 block 2 | ADC2, tile 225 block 0 |
| 3 | DAC7, tile 229 block 3 | ADC3, tile 225 block 1 |

XM500 silkscreen labels vary by board revision; tile and block identifiers are
the authoritative mapping. Antennas connect only to ADC-labelled ports.

RF output from a DAC requires a cabled and attenuated loopback path or an
otherwise compliant test arrangement. Applicable RF-emission and
signal-interception requirements remain in force.


## Software configuration

`rfsoc_sam.overlay.Overlay` programs the RFSoC-SAM bitstream and configures
the RF data converters and associated programmable-logic IP. Recorded overlay
initialisation time was approximately one minute.


```python
from rfsoc_sam.overlay import Overlay

sam = Overlay()
```

Overlay initialisation programs the ZCU111 LMK and LMX clock devices to
122.8 MHz and 409.6 MHz respectively. Recorded analyser-widget initialisation
time was approximately two minutes.


```python
analyzer = sam.spectrum_analyzer()
```

---


## Loopback tone control

The RFSoC-SAM tone-control widget sets RF-DAC frequency and amplitude. A cabled
loopback provides the spectrum-analyser stimulus.


```python
analyzer.children[2]
```

## Spectrum-analyser capability

- Maximum inspected bandwidth: 1638.4 MHz
- Minimum recorded spectral resolution: 0.244140625 kHz
- Adjustable bandwidth range: 1.6 MHz to 1638.4 MHz
- Tuning range: 0 MHz to 4096 MHz using higher Nyquist zones


```python
analyzer.children[1]
```

The programmable-logic FFT supports lengths from 64 to 8192 points. Subsequent
floating-point and decibel processing produces power-spectrum or
power-spectral-density display values. In loopback operation, the configured
RF-DAC tone appears in the corresponding RF-ADC spectrum.


## 2.4 GHz observation record

A 2400 MHz centre frequency and decimation factor of 16 provide a narrow-band
view of nearby 2.4 GHz activity. Local transmitters and test stimuli remain
subject to normal RF safety and regulatory controls.
